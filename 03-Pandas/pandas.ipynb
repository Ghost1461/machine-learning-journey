#### Gözlem ve Değişken Seçimi(loc, iloc)


```python
import numpy as np
import pandas as pd
m = np.random.randint(1,30, size = (10,3))
df = pd.DataFrame(m, columns = ["var1","var2","var3"])
print(df, "\n")
print(df[:1], "\n")
print(df["var1"]) #sütunu verir
```

```python
print(df.loc[0:3])
print(df.iloc[0:3])
```

```python
print(df.loc[0:3, "var3"])
print(df.iloc[0:4, 2])
```

```python
print(df, "\n")
print(df.iloc[0], "\n") #satırı verir
print(df["var1"], "\n") #var1'in hepsini verir
print(df["var1"][1], "\n") #var1'in 2. satırdaki değerini verir
print(df[0:2][["var1","var2"]]) #ilk iki satırdaki var1 ve var2 değerlerini verir
```

#### Dataframe Birleştirme (Concat)

```python
import numpy as np
import pandas as pd
m = np.random.randint(1,30, size = (5,3))
df = pd.DataFrame(m, columns = ["var1","var2","var3"])
print(df, "\n")
```

```python
df2 = df*5
df2
```

```python
pd.concat([df, df2], ignore_index = True, axis = 0)
```

```python
df2.columns = ["var1", "var2", "zorti"]
```

```python
pd.concat([df, df2], ignore_index = True, axis = 0, join = "inner") #ortakları al
```

```python
pd.concat([df, df2], ignore_index = True, axis = 0, join = "outer") #ikisinden gelen sütunlarıda al
```

```python
pd.concat([df, df2], ignore_index = True, axis = 0).reindex(columns=df2.columns) #df2'nin sütunlarına göre biçimlendir
```

#### İleri Birleştirme İşlemleri (Merge)

```python
df1 = pd.DataFrame({'calisanlar': ['Ali', 'Veli', 'Ayse', 'Fatma'],'grup': ['Muhasebe', 'Muhendislik', 'Muhendislik', 'İK']})

df1
```

```python
df2 = pd.DataFrame({'calisanlar': ['Ayse', 'Ali', 'Veli', 'Fatma'],'ilk_giris': [2010, 2009, 2014, 2019]})

df2
```

```python
pd.merge(df1, df2) #ortak column'u kendi otomatik anlar ve onun üzerinden birleştirir
```

```python
df3 = pd.merge(df1, df2, on = "calisanlar") #calisanlar column'u üzerinden birleştir
```

```python
df4 = pd.DataFrame({'grup': ['Muhasebe', 'Muhendislik', 'İK'],'mudur': ['Caner', 'Mustafa', 'Berkcan']})
df4
```

```python
pd.merge(df3, df4) #many to one merge
```

```python
df5 = pd.DataFrame({'grup': ['Muhasebe', 'Muhasebe',
                              'Muhendislik', 'Muhendislik', 'İK', 'İK'],
                    'yetenekler': ['matematik', 'excel', 'kodlama', 'linux',
                               'excel', 'yonetim']})

df5
```

```python
pd.merge(df1, df5) #many to many merge
```

#### Aggregate

```python
import pandas as pd
df = pd.DataFrame({'gruplar': ['A', 'B', 'C', 'A', 'B', 'C'],
                   'degisken1': [10,23,33,22,11,99],
                   'degisken2': [100,253,333,262,111,969]},
                   columns = ['gruplar', 'degisken1', 'degisken2'])
df
```

```python
print(df.groupby("gruplar").aggregate(["min", "median", "max"])) #istediğim istatistikleri getir
```

```python
print(df.groupby("gruplar").aggregate({"degisken1":"min", "degisken2":"count"})) #farklı columnlardan farklı istediğimiz şeyleri getirme
```

#### Filter(Koşula göre veri seçer)

```python
import pandas as pd
df = pd.DataFrame({'gruplar': ['A', 'B', 'C', 'A', 'B', 'C'],
                   'degisken1': [10,23,33,22,11,99],
                   'degisken2': [100,253,333,262,111,969]},
                   columns = ['gruplar', 'degisken1', 'degisken2'])
df
```

```python
def filter_func(x):
    return x["degisken1"].mean() > 20
```

```python
df.groupby("gruplar").mean()
```

```python
df.groupby("gruplar").filter(filter_func) #A, B, C  olarak gruplandıktan sonra degisken1 mean değeri 20'den büyük olan elemanları yazdırır
```

#### Transform(Verilerin boyutunu koruyarak dönüşüm yapar)

```python
import pandas as pd
df = pd.DataFrame({'gruplar': ['A', 'B', 'C', 'A', 'B', 'C'],
                   'degisken1': [10,23,33,22,11,99],
                   'degisken2': [100,253,333,262,111,969]},
                   columns = ['gruplar', 'degisken1', 'degisken2'])
df
```

```python
df_a = df.iloc[:,1:4]
df_a
```

```python
df_a.transform(lambda x: (x-x.mean())/x.std()) #x fonksiyona sırayla gönderdiğimiz her bir sütunu temsil eder.
'''
Pandas bunu her satıra ayrı ayrı fonksiyon çağırarak değil, 
sütunun tamamına tek seferde vektörel işlem uygulayarak gerçekleştiriyor.
'''
```

#### Apply(Bir fonksiyonu satırlara veya sütunlara uygular)

```python
import pandas as pd
df = pd.DataFrame({'degisken1': [10,23,33,22,11,99],
                   'degisken2': [100,253,333,262,111,969]},
                   columns = ['degisken1', 'degisken2'])
df
```

```python
df.apply(np.sum)
```

```python
df.apply(lambda x: x.mean(), axis=1) #her satıra mean uygular
```

```python
df = pd.DataFrame({
    "vize": [60, 80, 40],
    "final": [70, 90, 50]
})

df["ortalama"] = df.apply(
    lambda x: x["vize"] * 0.4 + x["final"] * 0.6,
    axis=1
)
print(df)
```

#### Pivot Tablolar

```python
import pandas as pd
import seaborn as sns
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
titanic.groupby("sex")["survived"].mean()
```

```python
titanic.groupby(["sex","class"])["survived"].aggregate("mean").unstack()
```

```python
titanic.pivot_table(values = "survived", index="sex", columns="class")
```

```python
age = pd.cut(titanic["age"], [0,18,90])
age.head(10)
```

```python
titanic.pivot_table(values = "survived", index=["sex", age], columns="class")
```
