#### Eleman İşlemleri

```python
import numpy as np
s1 = np.random.randint(10, size=5)
s2 = np.random.randint(10, size=5)
s3 = np.random.randint(10, size=5)
dict = {"var1":s1,"var2":s2,"var3":s3}
dict
```

```python
import pandas as pd
df = pd.DataFrame(dict)
print(df)
print(df.index)
df.index = ["a","b","c","d","e"]
print(df)
```

```python
dfg = df.drop("a", axis=0)
print(dfg)
print(df)
```

```python
df.drop("a", axis=0, inplace = True)
print(df)
```
