#### Concatenation: Array Birleştirme

###### Tek Boyutlu Array

```python
x = np.array([1,2,3])
y= np.array([4,5,6])
```

```python
np.concatenate([x,y])
```

```python
z=np.array([7,8,9])
np.concatenate([x,y,z])
```

###### İki Boyutlu Array

```python
a = np.array([[1,2,3],[4,5,6]])
a
```

```python
np.concatenate([a,a,a],axis=0)#Axis'in default değeri 0
```

```python
np.concatenate([a,a,a],axis=1)
```

#### Splitting: Array Ayırma

```python
x = np.array([1,2,3,99,3,99,3,2,1])
a,b,c = np.split(x,[2,6])#Verilen sayıya göre bölmelendirme
a
```

```python
m = np.arange(16).reshape(4,4)
m
```

```python
ust, alt = np.vsplit(m, [2]) #Yatay bölme
alt
```

```python
sol, sağ = np.hsplit(m, [2]) #Dikey bölme
sağ
```

#### Sorting: Sıralama

```python
import numpy as np
v = np.array([1,20,3,4,5])
v
```

```python
np.sort(v)#veri setinin orijinal yapısıyla oynamaz
```

```python
v.sort()#veri setinin kalıcı olarak değiştirir
v
```

```python
t = np.random.normal(20,5,(3,3))
t
```

```python
np.sort(t, axis=0)#kendi içinde sütunları sıralar
```

```python
np.sort(t, axis=1)#kendi içinde satırları sıralar
```

#### Index ile Erişim

```python
import numpy as np
a = np.random.randint(10, size=10)
print(a)
print(a[0])
print(a[-1])
a[0] = 3
print(a[0])
print(a)
```

```python
import numpy as np
m = np.random.randint(10, size=(3,5))
print(m)
print(m[0,0])
print(m[1,1])
print(m[1,4])
m[1,4] = 61
print(m)
```

#### Slicing

##### Tek Boyutlu Slicing

```python
import numpy as np
z = np.arange(20,30)
print(z[0:3])
print(z[0:])
print(z[:3])
print(z[0::3])#ilk elemandan başlayarak ikişer artarak git
```
