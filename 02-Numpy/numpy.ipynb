#### Index ile Erişim

```python
import numpy as np
a = np.random.randint(10, size=10)
print(a)
print(a[0])
print(a[-1])
a[0] = 3
print(a[0])
print(a)
```

```python
import numpy as np
m = np.random.randint(10, size=(3,5))
print(m)
print(m[0,0])
print(m[1,1])
print(m[1,4])
m[1,4] = 61
print(m)
```

#### Slicing

##### Tek Boyutlu Slicing

```python
import numpy as np
z = np.arange(20,30)
print(z[0:3])
print(z[0:])
print(z[:3])
print(z[0::3])#ilk elemandan başlayarak ikişer artarak git
```

##### Çift Boyutlu Slicing

```python
import numpy as np
m = np.random.randint(10, size = (5,5))
m
```

```python
print(m[:,0])
print(m[:,1])
print(m[:,4])
print(m[4,:])
print(m[0,4])
print(m[0:2,2:4])
```

#### Alt Küme Üzerinde İşlemler(copy)

```python
import numpy as np
a = np.random.randint(10, size = (5,5))
a
```

```python
alt_a = a[0:3, 0:2]
alt_a
```

```python
alt_a[0,0] = 6161
alt_a[1,1] = 8888
print(alt_a)
print(a)#alt kümesinde yaptığımız değişiklik ana array'i de etkiledi
```

##### copy()

```python
import numpy as np
b = np.random.randint(10, size = (5,5))
b
```

```python
alt_b = b[0:3, 0:2].copy()
alt_b
```

```python
alt_b[0,0] = 6161
print(alt_b)
print(b)
```

#### Fancy Index


##### Tek Boyutta

```python
import numpy as np
a = np.arange(0,30, 3)
print(a)
print([a[0],a[-1],a[3]])
```

```python
al_getir = [0,-1,3]
print(a[al_getir])
```

##### Çift Boyutta

```python
r = np.arange(9).reshape(3,3)
print(r)
satır = np.array([1,2])
sütun = np.array([1,2])
print(r[satır, sütun])#satır ve sütunu kendi eşleştirdi ve matrixdeki indexte bulunan değeri getirdi
```

```python
print(r[0, [1,2]])#ilk satırın 1. ve 2. indexinin getir
```

```python
print(r[0:, [1,2]])#tüm satırların 1. ve 2. indexini
```

```python
print(r[[1,2],[0,0]])#ilk köşeli parantez satırlar, ikinci sütunlar; çalıştırınca eşlenirler
```
